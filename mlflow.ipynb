```python
import pandas as pd 
import numpy as np 
```

```python
from sklearn.datasets import load_wine
```

```python
wine = load_wine()
x = wine.data
y = wine.target
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score , confusion_matrix
import mlflow
import mlflow.sklearn
```

```python
x_train , x_test ,y_train , y_test = train_test_split(x,y ,test_size=0.2,random_state=42)
```

```python
parameter defined 
```

```python
n_estimators = 10 
max_depth = 5 
```

```python
mlflow.set_experiment('wine_datasets')
from sklearn.metrics import confusion_matrix

with mlflow.start_run():
    model = RandomForestClassifier(n_estimators=n_estimators , max_depth=max_depth )

    model.fit(x_train , y_train)
    y_pre = model.predict(x_test)
    acc = accuracy_score(y_test , y_pre)
    confusion = confusion_matrix(y_test , y_pre)

    mlflow.log_param('n_estimators' ,n_estimators)
    mlflow.log_param('max_depth' , max_depth)

    mlflow.log_metric('accuracy' ,acc)
    mlflow.log_metric('confusion_metrix' , confusion)

    mlflow.sklearn.load_model('RandomForest' ,model)
    
```

```python
import mlflow
import mlflow.sklearn
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import confusion_matrix, accuracy_score, ConfusionMatrixDisplay
from sklearn.ensemble import RandomForestClassifier

# Set experiment
mlflow.set_experiment("wine_datasets")

with mlflow.start_run():

    model = RandomForestClassifier(
        n_estimators=n_estimators,
        max_depth=max_depth,
        random_state=42
    )

    model.fit(x_train, y_train)

    y_pred = model.predict(x_test)

    acc = accuracy_score(y_test, y_pred)
    cm = confusion_matrix(y_test, y_pred)

    # Log params
    mlflow.log_param('n_estimators', n_estimators)
    mlflow.log_param('max_depth', max_depth)

    # Log metric (ONLY scalar)
    mlflow.log_metric('accuracy', acc)

    # ✅ Save confusion matrix as image
    disp = ConfusionMatrixDisplay(confusion_matrix=cm)
    disp.plot()
    plt.savefig("confusion_matrix.png")
    mlflow.log_artifact("confusion_matrix.png")
    plt.close()

    # ✅ Log model correctly
    mlflow.sklearn.log_model(model, "RandomForest")
```
